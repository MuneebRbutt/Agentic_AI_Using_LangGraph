```python
from langgraph.graph import START, END, StateGraph
from dotenv import load_dotenv
from typing import TypedDict, Literal
from langchain_openai import ChatOpenAI
from pydantic import BaseModel, Field

```

```python
load_dotenv()
```

```python
model = ChatOpenAI(model = "gpt-4o-mini")
```

```python
class SentimentSchema(BaseModel):
    sentiment: Literal["positive", "negative"] = Field(description= "Sentiment of the review")
    
```

```python
class DiagnosisSchema(BaseModel):
    issue_type: Literal["UX", "Performance","Bug", "Support","Other"] = Field(description="Sentiment of the review")
    tone: Literal["Calm", "Angry", "Frustrated", "Dissappointed"] = Field(description= "The emotional state of the user")
    urgency: Literal["low", "medium", "high"] = Field(description="How urgent and critical is the issue.")
    
    
```

```python
structured_model = model.with_structured_output(SentimentSchema)
structured_model_diagnosis= model.with_structured_output(DiagnosisSchema)


review = 'The software quality was terrible, its battery was worse'

prompt = f'What is the sentiment of the following review-> {review}'
```

```python
structured_model.invoke(prompt).sentiment
```

```python
class ReviewState(TypedDict):
    review: str
    sentiment: Literal["positive", "negative"]
    diagnosis: str
    response: str
```

```python
def find_sentiment(state: ReviewState):
    
    prompt = f'For the following review, find out the sentiment \n {state["review"]}'
    sentiment = structured_model.invoke(prompt).sentiment
    
    return {'sentiment': sentiment}

def check_sentiment(state: ReviewState) -> Literal["positive_response", "run_diagnosis"]:
    if state['sentiment'] == "positive":
        return "positive_response"
    else:
        return "run_diagnosis"
    
def positive_response(state: ReviewState):
    prompt = f"""Give a thankyou reponse to the user based on this review. \n {state['review']}\n Also ask them to give the feedback on our website as well""" 
    
    response = model.invoke(prompt).content
    
    return {'response': response}  



def run_diagnosis(state: ReviewState):
    prompt = f"""Diagnose this negative review:\n\n {state['review']}
    Mention the urgency, tone, issue_type
    """
    response = structured_model_diagnosis.invoke(prompt)
    
    return {'diagnosis': response.model_dump()}

def negative_response(state: ReviewState):
    
    diagnosis = state['diagnosis']
    
    prompt = f"""You are helpful assistant, with this issue\n {diagnosis['issue_type']}, urgency is marked as \n {diagnosis['urgency']}, with this tone {diagnosis['tone']}. Write an emphathetic, helpful resoution message"""
    
    response = model.invoke(prompt).content
    
    return {'response': response} 


```

```python
graph = StateGraph(ReviewState)

graph.add_node('find_sentiment', find_sentiment)
graph.add_node('positive_response', positive_response)
graph.add_node('negative_response', negative_response)
graph.add_node('run_diagnosis', run_diagnosis)

graph.add_edge(START, 'find_sentiment')
graph.add_conditional_edges('find_sentiment', check_sentiment)
graph.add_edge('positive_response', END)
graph.add_edge('run_diagnosis', 'negative_response')
graph.add_edge('negative_response', END)

workflow = graph.compile()
workflow
```

```python
initial_state = {
    'review': """I’m quite disappointed with this laptop. The performance is not as smooth as expected, and it tends to slow down during multitasking or when running heavier applications. The battery life is also below average, and the laptop can get noticeably hot after extended use.

The build quality is acceptable, but considering the price, I expected a much better overall experience. There are better options available in the same price range, so I wouldn’t personally recommend this laptop."""
}

workflow.invoke(initial_state)
```
